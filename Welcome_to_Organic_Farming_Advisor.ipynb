```python
import gradio as gr

# Update Seasons
def update_season(crop):# this function runs when ever user select a crop
    if crop == "Paddy":
        return gr.update(choices=["Kharif", "Rabi"], value=None)

    elif crop == "Wheat":
        return gr.update(choices=["Kharif", "Rabi"], value=None)

    elif crop == "Cotton":
        return gr.update(choices=["Kharif", "Rabi"], value=None)

    elif crop == "Chilli":
        return gr.update(choices=["Kharif", "Rabi"], value=None)

    else:
        return gr.update(choices=[], value=None)


#  Update Seeds
def update_seed(crop, season):# this function checks the both selected crop and season then display the available seeds types

    # Paddy
    if crop == "Paddy" and season == "Kharif":
        return (
            gr.update(choices=["MTU-1010", "BPT-5204"], value=None),
            ""
        )

    elif crop == "Paddy" and season == "Rabi":
        return (
            gr.update(choices=["RNR-15048", "MTU-7029"], value=None),
            ""
        )

    # Wheat
    elif crop == "Wheat" and season == "Kharif":
        return (
            gr.update(choices=[], value=None),
            " Wheat cannot be grown in Kharif season because it requires a cool climate."
        )

    elif crop == "Wheat" and season == "Rabi":
        return (
            gr.update(choices=["DBW-327", "MACS-6222"], value=None),
            ""
        )

    # Cotton
    elif crop == "Cotton" and season == "Kharif":
        return (
            gr.update(choices=["RASI-659", "GOLD-224"], value=None),
            ""
        )

    elif crop == "Cotton" and season == "Rabi":
        return (
            gr.update(choices=["KAVERI-2059", "NUZIVEEDU-5963"], value=None),
            ""
        )

    # Chilli
    elif crop == "Chilli" and season == "Kharif":
        return (
            gr.update(choices=["TEJA CHILLI", "GUNTUR SANAM"], value=None),
            ""
        )

    elif crop == "Chilli" and season == "Rabi":
        return (
            gr.update(choices=["GREEN CHILLI", "RED CHILLI"], value=None),
            ""
        )

    else:
        return (
            gr.update(choices=[], value=None),
            ""
        )


#  Crop Duration
def show_duration(crop, season, seed): # when user select seed type this function run and display crop duration

    if crop == "Paddy" and seed == "MTU-1010":
        return "Crop Duration : 100-120 Days"

    elif crop == "Paddy" and seed == "BPT-5204":
        return "Crop Duration : 135-145 Days"

    elif crop == "Paddy" and seed == "RNR-15048":
        return "Crop Duration : 125-135 Days"

    elif crop == "Paddy" and seed == "MTU-7029":
        return "Crop Duration : 115-125 Days"

    elif crop == "Wheat" and seed == "DBW-327":
        return "Crop Duration : 120-130 Days"

    elif crop == "Wheat" and seed == "MACS-6222":
        return "Crop Duration : 110-120 Days"

    elif crop == "Cotton" and seed == "RASI-659":
        return "Crop Duration : 160-180 Days"

    elif crop == "Cotton" and seed == "GOLD-224":
        return "Crop Duration : 150-170 Days"

    elif crop == "Cotton" and seed == "KAVERI-2059":
        return "Crop Duration : 140-160 Days"

    elif crop == "Cotton" and seed == "NUZIVEEDU-5963":
        return "Crop Duration : 150-170 Days"

    elif crop == "Chilli" and seed == "TEJA CHILLI":
        return "Crop Duration : 140-160 Days"

    elif crop == "Chilli" and seed == "GUNTUR SANAM":
        return "Crop Duration : 150-170 Days"

    elif crop == "Chilli" and seed == "GREEN CHILLI":
        return "Crop Duration : 90-110 Days"

    elif crop == "Chilli" and seed == "RED CHILLI":
        return "Crop Duration : 120-140 Days"

    else:
        return ""


#  Problems
def update_problem(crop): #this function is used to update the list of problems relatrd to that crop

    if crop == "Paddy":
        return gr.update(
            choices=["Yellow Leaves", "Stem Borer", "Low Growth"],
            value=None
        )

    elif crop == "Wheat":
        return gr.update(
            choices=["Root Rot", "Yellow Leaves", "Weak Growth"],
            value=None
        )

    elif crop == "Cotton":
        return gr.update(
            choices=["Bollworm", "Whitefly", "Poor Growth"],
            value=None
        )

    elif crop == "Chilli":
        return gr.update(
            choices=["Leaf Curl", "Thrips", "Fruit Rot"],
            value=None
        )

    else:
        return gr.update(choices=[], value=None)


#  Organic Solution
def organic_solution(crop, problem): #this function gives the organic solution for the selected crop problem

    if crop == "Paddy" and problem == "Yellow Leaves":
        return "Apply Jeevamrutham and Panchagavya spray every 15 days."

    elif crop == "Paddy" and problem == "Stem Borer":
        return "Use Neem oil spray (5 ml/L) and install pheromone traps."

    elif crop == "Paddy" and problem == "Low Growth":
        return "Apply vermicompost and fermented cow dung solution."

    elif crop == "Wheat" and problem == "Root Rot":
        return "Apply Trichoderma mixed with compost."

    elif crop == "Wheat" and problem == "Yellow Leaves":
        return "Use liquid biofertilizer and seaweed extract spray."

    elif crop == "Wheat" and problem == "Weak Growth":
        return "Apply vermiwash and farmyard manure."

    elif crop == "Cotton" and problem == "Bollworm":
        return "Use Neem Seed Kernel Extract (NSKE 5%)."

    elif crop == "Cotton" and problem == "Whitefly":
        return "Spray Neem oil and maintain field sanitation."

    elif crop == "Cotton" and problem == "Poor Growth":
        return "Apply compost tea and Jeevamrutham."

    elif crop == "Chilli" and problem == "Leaf Curl":
        return "Spray Neem oil and remove infected plants."

    elif crop == "Chilli" and problem == "Thrips":
        return "Use sticky traps and garlic-chilli extract."

    elif crop == "Chilli" and problem == "Fruit Rot":
        return "Spray Trichoderma solution."

    else:
        return "Please select a valid crop and problem."


# Gradio UI
with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("# 🌱 Welcome to Organic Farming Advisor 🌱 ") # project title

    crop = gr.Dropdown(
        label="Select Crop",
        choices=["Paddy", "Wheat", "Cotton", "Chilli"]
    ) # it allows the user to select one crop

    season = gr.Dropdown(label="Select Season", choices=[])# initially choices=[] means the dropdown is empth becouse the season depends on selected crop

    seed = gr.Dropdown(label="Select Seed Type", choices=[])

      # Message Box
    message = gr.Textbox(
        label="Message",
        interactive=False
    )


    duration = gr.Textbox(
        label="Crop Duration",
        interactive=False # means user can not change output
    )

    problem = gr.Dropdown(
        label="Select Problem",
        choices=[]
    )

    submit = gr.Button("Submit", variant="primary")

    solution = gr.Textbox(
        label="Organic Solution",
        lines=4,
        interactive=False
    )

    # Events (an event tells gradio when to run a function )
    crop.change(update_season, crop, season)

    season.change(
        update_seed,
        inputs=[crop, season],
        outputs=[seed, message]
    )

    seed.change(
        show_duration,
        inputs=[crop, season, seed],
        outputs=duration
    )

    crop.change(
        update_problem,
        inputs=crop,
        outputs=problem
    )

    submit.click(
        organic_solution,
        inputs=[crop, problem],
        outputs=solution
    )

demo.launch()
```
